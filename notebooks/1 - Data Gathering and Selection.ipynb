### Exercise 1.1

Explore the main 3-5 categories of master agreements over time to confirm they have a similar distribution over the years.

```python
# 1. Filter for Master Agreements only
master_agreements = contracts[(pd.isna(contracts['Contract PDF'])==False) & 
                              (contracts['Procurement Type']=='MASTER AGREEMENT')]

# 2. Identify the top 5 Departments (Categories)
top_departments = master_agreements['Department'].value_counts().head(5).index.tolist()

# 3. Plot distribution over time for each top department
plt.figure(figsize=(10, 6))

for dept in top_departments:
    # Filter for this specific department
    dept_data = master_agreements[master_agreements['Department'] == dept]
    
    # Convert date and get year counts
    dates = pd.to_datetime(dept_data['Approval Date'])
    year_counts = dates.dt.year.value_counts().sort_index()
    
    # Plot
    year_counts.plot(kind='line', label=dept, marker='o')

plt.title('Top 5 Master Agreement Categories Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Contracts')
plt.legend(bbox_to_anchor=(1.5, 1), loc='upper right')
plt.grid(True)
plt.show()
```

## Download Selected Sample

Now, we need to create a quick RPA tool so that we don't have to manually go to each URL and download the documents. Due to the nature of the website, we'll leverage Selenium. By configuring Selenium, we can automatically download the documents, given the URL link in the metadata file:

```python
def download_pdf(lnk: str, download_folder: str):
    
    options = webdriver.ChromeOptions()
    profile = {"download.default_directory": os.path.abspath(download_folder),
               "download.prompt_for_download": False,
               "plugins.always_open_pdf_externally": True}
    options.add_experimental_option("prefs", profile)            
    
    # Windows Adjustment: Service for Chrome 133+
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
    
    try:
        driver.get(lnk)
        print(f"Downloading: {lnk}")
        
       
        wait = WebDriverWait(driver, 20)
        wait.until(EC.frame_to_be_available_and_switch_to_it((By.XPATH,"//iframe[@id='pdfFrame']")))
        wait.until(EC.element_to_be_clickable((By.ID, "open-button"))).click()    
                
        timeout = 60  
        start_time = time.time()
        print("Waiting for download to finish...")
        
        while time.time() - start_time < timeout:
            time.sleep(2)
            # Check if any file in the folder is still downloading
            downloading = [f for f in os.listdir(download_folder) if f.endswith('.crdownload')]
            if not downloading:
                # Check if a new PDF has actually appeared
                pdfs = [f for f in os.listdir(download_folder) if f.endswith('.pdf') and "DPSWeb" in f]
                if pdfs:
                    break
        
        filename = lnk.split("/")[-1].split('=')[-1] + '.pdf'
        all_pdfs = [os.path.join(download_folder, f) for f in os.listdir(download_folder) if f.endswith('.pdf')]
        
        if all_pdfs:
            latest_file = max(all_pdfs, key=os.path.getmtime)
            target_path = os.path.join(download_folder, filename)
            
            if os.path.exists(target_path):
                os.remove(target_path) # Remove old version if it exists
            os.rename(latest_file, target_path)
            print(f"Status: Success. Saved as {filename}")
        else:
            print("Status: Error - No PDF found.")

    finally:
        driver.quit() # Always use quit() on Windows
```

Let's pick now take our sample and download - it will take some time to run over our selected sample. We'll save the downloaded PDFs to the "data_raw" folder in our repo. Alternatively, you can populate the "data_raw" folder from the folder provided on Google Drive.

```python
download_folder = os.path.join(HOME_DIRECTORY, "data", "data_raw")

files0 = contracts[(pd.isna(contracts['Contract PDF'])==False)&(contracts['Procurement Type']=='MASTER AGREEMENT')]['Contract PDF'].values
files1 = contracts[(pd.isna(contracts['Contract PDF'])==False)&(contracts['Procurement Type']=='BID')]['Contract PDF'].sample(500, random_state=123)
files2 = contracts[(pd.isna(contracts['Contract PDF'])==False)&(contracts['Procurement Type']=='RFP')]['Contract PDF'].sample(300, random_state=234)
files3 = contracts[(pd.isna(contracts['Contract PDF'])==False)&(contracts['Procurement Type']=='RFQ')]['Contract PDF'].sample(200, random_state=345)
files = np.concatenate((files0, files1, files2, files3))

for i, f in enumerate(files):
    print(f"--- Processing index {i} ---")
    try:
        download_pdf(f, download_folder)
    except OSError as err:
        print("OS error:", err)
    except ValueError as err:
        print("URL or folder is invalid:", err)
    except Exception as err:
       
        print(f"Skipping index {i} due to unexpected error: {err}")

print("Batch processing finished.")
```
